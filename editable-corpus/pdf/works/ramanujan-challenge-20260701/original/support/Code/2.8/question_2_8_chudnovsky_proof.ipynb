## 4. Normalized tails and derivation of the initial matrix

```python
z0=-q; A0=sp.Integer(545140134); B0=sp.Integer(13591409); C0=sp.Integer(426880)
def H4(xs,ys):
    X=sp.symbols('X'); poly=sp.Poly(sp.expand(X*sp.prod(X+y-1 for y in ys)-z0*sp.prod(X+x for x in xs)),X); a=poly.all_coeffs(); a=[sp.simplify(v/a[0]) for v in a]; _,d3,d2,d1,d0=a
    return sp.Matrix([[0,0,0,-d0],[1,0,0,-d1],[0,1,0,-d2],[0,0,1,-d3]])
def up(xs,ys,i):
    M=sp.eye(4)+H4(xs,ys)/xs[i]; xx=list(xs);xx[i]+=1;return M,tuple(xx),ys
def low(xs,ys,j):
    yy=list(ys);yy[j]+=1; M=(sp.eye(4)+H4(xs,tuple(yy))/ys[j]).inv();return M,xs,tuple(yy)
xs=(sp.Integer(1),sp.Rational(7,6),sp.Rational(3,2),sp.Rational(11,6));ys=(2,2,2);T12=sp.eye(4)
for i in [1,2,3]: M,xs,ys=up(xs,ys,i);T12=sp.simplify(T12*M)
for j in [0,1,2]: M,xs,ys=low(xs,ys,j);T12=sp.simplify(T12*M)
rho1=z0*sp.Rational(1,6)*sp.Rational(1,2)*sp.Rational(5,6)
rho2=z0*sp.Rational(7,6)*sp.Rational(3,2)*sp.Rational(11,6)/8
e0=sp.Matrix([1,0,0,0]); r=e0-rho2*T12*e0; u1=sp.Matrix([A0+B0,A0,0,0]); alpha=sp.simplify(B0*r+rho1*u1); beta=sp.simplify(C0*r)
ICraw=sp.Matrix.vstack(alpha.T,beta.T)
Achallenge=sp.Matrix([[37169305760442252761441,111507917281327441564208,111507917281327599720129,37169305760442410917362],[1167416361542639692320,3502249084627896132160,3502249084627879697280,1167416361542622723840]])
scale=sp.Integer(2734764715008000)
assert ICraw*scale==Achallenge
print('PASS: tail-derived alpha,beta and exact scaling of the challenge initial matrix')
```

```python
try:
    from ramanujantools import Position
    from ramanujantools.cmf.pfq import pFq
except ImportError as exc:
    print(
        'RamanujanTools reconstruction skipped in this environment:',
        type(exc).__name__,
        exc,
    )
else:
    x0, x1, x2, x3, y0, y1, y2 = sp.symbols(
        'x0 x1 x2 x3 y0 y1 y2'
    )

    def normalize_rt_matrix(A):
        """Rewrite RamanujanTools' internal symbol named n as this notebook's n."""
        A = sp.Matrix(A)
        replacements = {s: n for s in A.free_symbols if s.name == 'n'}
        if replacements:
            A = A.xreplace(replacements)
        return A.applyfunc(sp.cancel)

    cmf_rt = pFq(4, 3, -sp.Rational(1, 53360**3))
    direction_rt = Position({
        x0: 1, x1: 1, x2: 1, x3: 1,
        y0: 2, y1: 2, y2: 2,
    })
    initial_rt = Position({
        x0: 1,
        x1: sp.Rational(7, 6),
        x2: sp.Rational(3, 2),
        x3: sp.Rational(11, 6),
        y0: 2, y1: 2, y2: 2,
    })

    # Reconstruct every core matrix returned by RamanujanTools in the helper.
    T_rt_raw = cmf_rt.trajectory_matrix(direction_rt, initial_rt)
    C_rt_raw = T_rt_raw.as_companion()
    K_rt_raw = T_rt_raw.companion_coboundary_matrix()
    TinvT_rt_raw = T_rt_raw.inv().T
    Mpoly_rt_raw = TinvT_rt_raw.as_polynomial()

    T_rt = normalize_rt_matrix(T_rt_raw)
    C_rt = normalize_rt_matrix(C_rt_raw)
    K_rt = normalize_rt_matrix(K_rt_raw)
    TinvT_rt = normalize_rt_matrix(TinvT_rt_raw)
    Mpoly_rt = normalize_rt_matrix(Mpoly_rt_raw)

    assert zero_matrix(T_rt - T), (
        'RamanujanTools 1111222 trajectory differs from T.'
    )
    assert zero_matrix(C_rt - C), (
        'RamanujanTools companion differs from C.'
    )
    assert zero_matrix(K_rt - K), (
        'RamanujanTools companion coboundary differs from K.'
    )
    assert zero_matrix(
        T_rt * K_rt.subs(n, n + 1) - K_rt * C_rt
    ), 'RamanujanTools companion coboundary identity failed.'


    assert zero_matrix(TinvT_rt - TinvT), (
        'RamanujanTools inverse transpose differs from TinvT.'
    )
    assert zero_matrix(Mchallenge - sigma * TinvT_rt), (
        'The package inverse transpose does not reproduce the challenge matrix.'
    )

    # The helper also calls as_polynomial() on the inverse transpose.  Its
    # normalization need not be the challenge normalization, so check exact
    # projective proportionality over Q(n), not literal equality.
    poly_scale = sp.cancel(Mpoly_rt[0, 0] / Mchallenge[0, 0])
    assert poly_scale != 0
    assert zero_matrix(Mpoly_rt - poly_scale * Mchallenge), (
        'RamanujanTools as_polynomial() is not proportional to Mchallenge.'
    )

    print(
        'PASS: RamanujanTools reconstructs all 1111222 proof matrices: '
        'trajectory, companion matrix, companion coboundary, inverse '
        'transpose, polynomial normalization, challenge normalization, '
        'and the exact coboundary identity.'
    )
```
