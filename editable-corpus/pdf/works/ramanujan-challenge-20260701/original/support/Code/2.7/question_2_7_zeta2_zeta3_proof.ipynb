## 7. RamanujanTools reconstruction of the primal and dual companion data

The final cell reproduces the helper-notebook workflow in full.  Starting from the ${}_4F_3$ CMF, it reconstructs the primal trajectory, its companion matrix and companion coboundary, then repeats `as_companion()` and `companion_coboundary_matrix()` for the inverse-transpose trajectory.  All six package-generated matrices are compared entry by entry with the exact matrices used in this notebook.  Import failure is the only skipped case.


```python
try:
    from ramanujantools import Position
    from ramanujantools.cmf.pfq import pFq
except ImportError as exc:
    print(
        'RamanujanTools reconstruction skipped in this environment:',
        type(exc).__name__,
        exc,
    )
else:
    x0, x1, x2, x3, y0, y1, y2 = sp.symbols(
        'x0 x1 x2 x3 y0 y1 y2'
    )

    def normalize_rt_matrix(A):
        """Rewrite RamanujanTools' internal symbol named n as this notebook's n."""
        A = sp.Matrix(A)
        replacements = {s: n for s in A.free_symbols if s.name == 'n'}
        if replacements:
            A = A.xreplace(replacements)
        return A.applyfunc(sp.cancel)

    cmf_rt = pFq(4, 3, 1)
    direction_rt = Position({
        x0: 1, x1: 1, x2: 1, x3: 1,
        y0: 2, y1: 2, y2: 2,
    })
    initial_rt = Position({
        x0: 1, x1: 1, x2: 1, x3: 1,
        y0: 1, y1: 1, y2: 2,
    })

    # Primal trajectory and every primal companion object returned by RT.
    T_rt_raw = cmf_rt.trajectory_matrix(direction_rt, initial_rt)
    Cprimal_rt_raw = T_rt_raw.as_companion().factor()
    V_rt_raw = T_rt_raw.companion_coboundary_matrix()

    T_rt = normalize_rt_matrix(T_rt_raw)
    Cprimal_rt = normalize_rt_matrix(Cprimal_rt_raw)
    V_rt = normalize_rt_matrix(V_rt_raw)

    assert zero_matrix(T_rt - T), (
        'RamanujanTools primal trajectory differs from T.'
    )
    assert zero_matrix(Cprimal_rt - Cprimal), (
        'RamanujanTools primal companion differs from Cprimal.'
    )
    assert zero_matrix(V_rt - V), (
        'RamanujanTools primal companion coboundary differs from V.'
    )
    assert zero_matrix(
        T_rt * V_rt.subs(n, n + 1) - V_rt * Cprimal_rt
    ), 'RamanujanTools primal coboundary identity failed.'
    assert Cprimal_rt[:, 0] == sp.Matrix([0, 1, 0])
    assert Cprimal_rt[:, 1] == sp.Matrix([0, 0, 1])

    # Inverse-transpose trajectory and every dual companion object returned by RT.
    M_rt_raw = T_rt_raw.inv().T
    Cdual_rt_raw = M_rt_raw.as_companion().factor()
    K_rt_raw = M_rt_raw.companion_coboundary_matrix()

    M_rt = normalize_rt_matrix(M_rt_raw)
    Cdual_rt = normalize_rt_matrix(Cdual_rt_raw)
    K_rt = normalize_rt_matrix(K_rt_raw)

    assert zero_matrix(M_rt - M), (
        'RamanujanTools inverse-transpose trajectory differs from M.'
    )
    assert zero_matrix(Cdual_rt - Cdual), (
        'RamanujanTools dual companion differs from Cdual.'
    )
    assert zero_matrix(K_rt - K), (
        'RamanujanTools dual companion coboundary differs from K.'
    )
    assert zero_matrix(
        M_rt * K_rt.subs(n, n + 1) - K_rt * Cdual_rt
    ), 'RamanujanTools dual coboundary identity failed.'


    print(
        'PASS: RamanujanTools reconstructs all six 2.7 matrices exactly: '
        'T, C_primal, V, T^{-T}, C_dual, K, together with both '
        'companion-coboundary identities.'
    )
```
